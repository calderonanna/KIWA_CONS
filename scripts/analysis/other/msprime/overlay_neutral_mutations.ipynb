# Overlay Neutral Mutations
A common approach to increase the efficiency of simulations is to combine SLiM forward-in-time simulations with backwards-in-time coalescent simulators like msprime. Simulating neutral mutations in SLiM is very expensive, so I ran my burn-ins and main simulations with SLiM and then use msprime to overlay neutral mutations. 

**An important clarification that I got confused with**: Recaptiation in msprime is the process of finishing any uncoalesced lineages. I let all of my burn-ins run until they were fully coalesced, I do not need recapitation. 

## Helpful Resources:
- SLiM 5 Manual Section 18.2 (Pg.554)
- tskit and msprime manual: https://tskit.dev/msprime/docs/stable/mutations.html

## Import Packages, Modify Data, Overly Neutral Mutations
-Select the msprime environment in the kernel options (created in the previous section). 
-Use `simplify()` to overaly mutations only back to the point of coalescence. Since I am not interested in fixed neutral mutations (e.g, mutations that are shared in all present-day samples), I will be using `simplify()`. 
-After `msprime` overlays neutral mutations and then generates a new tree sequence.

```python
#Import Packages
import msprime, tskit
from IPython.display import SVG

#Import Data and Simplify
ts = tskit.load("/Users/abc6435/Desktop/treeseqfinal_1000.trees")
ts = ts.simplify()

#Overlay Neutral Mutations
mutated = msprime.sim_mutations(ts, rate=4e-9, random_seed=1, keep=True)
```

## Calculate Genetic Metrics

```python
print(mutated.diversity())
```
